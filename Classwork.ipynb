```python
# Write a number guessing game where the user guesses a number between 1 and 10.
# Modify the guessing game to keep asking until the correct number is guessed.
# Add a counter to the guessing game to track the number of attempts.
```

```python
correct = 7
attempts = 0
while True:
    number = int(input('Guess a number between 1 and 10: '))
    if number == correct:
        print('You guessed right!')
        break
    else:
        print('try again')
        attempts += 1
print(f'you attempted {attempts} times')
```
