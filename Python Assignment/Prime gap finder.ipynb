```python
# Prime check function
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True
```

```python
# Generate all primes <10,000
primes = []

for num in range(2, 10001):
    if is_prime(num):
        primes.append(num)
```

```python
# Find consecutive prime gaps >= 20
prime_gaps = []

for i in range(len(primes) - 1):
    p1 = primes[i]
    p2 = primes[i + 1]
    gap = p2 - p1

    if gap >= 20:
        prime_gaps.append((p1, p2, gap))
```

```python
# Save result to a csv file
import csv

with open("prime_gaps.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Prime 1", "Prime 2", "Gap"])

    for p1, p2, gap in prime_gaps:
        writer.writerow([p1, p2, gap])
```
