```python
import random

high_scores = []
current_streak = 0
difficulty_range = (1, 100)


def adjust_difficulty():
    """Adjust difficulty range based on player streak."""
    global difficulty_range
    
    if current_streak >= 3:
        difficulty_range = (1, 200)
        return "🔥 Difficulty increased! Range: 1-200"
    elif current_streak <= -2:
        difficulty_range = (1, 50)
        return "💚 Difficulty decreased! Range: 1-50"
    else:
        difficulty_range = (1, 100)
        return "Range: 1-100"


def update_high_scores(name, attempts):
    """Update top-5 high scores."""
    global high_scores
    high_scores.append((name, attempts))
    high_scores = sorted(high_scores, key=lambda x: x[1])[:5]


def display_high_scores():
    """Display top-5 high scores."""
    if not high_scores:
        return "No high scores yet!"
    
    result = "\n🏆 TOP 5 HIGH SCORES 🏆\n"
    result += "="*30 + "\n"
    
    for i, (name, attempts) in enumerate(high_scores, 1):
        medal = "🥇" if i == 1 else "🥈" if i == 2 else "🥉" if i == 3 else "  "
        result += f"{medal} {i}. {name:<15} {attempts} attempts\n"
    
    return result


def play_round():
    """Play one round of the guessing game."""
    global current_streak
    
    print(f"\n{adjust_difficulty()}")
    secret = random.randint(difficulty_range[0], difficulty_range[1])
    attempts = 0
    
    print(f"I'm thinking of a number between {difficulty_range[0]} and {difficulty_range[1]}!")
    
    while True:
        try:
            guess = input("Your guess (or 'q' to quit): ").strip()
            
            if guess.lower() == 'q':
                print(f"The number was {secret}")
                return None
            
            guess = int(guess)
            attempts += 1
            
            if guess < secret:
                print("Too low!")
            elif guess > secret:
                print("Too high!")
            else:
                print(f"🎉 Correct! You got it in {attempts} attempts!")
                current_streak += 1
                return attempts
                
        except ValueError:
            print("Please enter a valid number or 'q' to quit")


def play_game():
    """Main game loop."""
    print("="*50)
    print("ADAPTIVE GUESSING GAME")
    print("="*50)
    print("Guess the number! Your performance affects difficulty.")
    
    while True:
        attempts = play_round()
        
        if attempts is None:
            print("\nThanks for playing!")
            break
        
        name = input("\nEnter your name for the leaderboard: ").strip()
        if name:
            update_high_scores(name, attempts)
        
        print(display_high_scores())
        
        play_again = input("\nPlay again? (y/n): ").strip().lower()
        if play_again != 'y':
            print("\nThanks for playing!")
            print(display_high_scores())
            break


# To play the game, uncomment the line below:
# play_game()

# Demo (simulated rounds)
print("=== Adaptive Guessing Game Demo ===\n")
print("Game mechanics:")
print("- Start with range 1-100")
print("- Win 3+ times: difficulty increases to 1-200")
print("- Struggle: difficulty decreases to 1-50")
print("- Top 5 scores are saved")
print("\nTo play interactively, uncomment: play_game()")
```
