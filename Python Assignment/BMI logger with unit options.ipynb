```python

from datetime import datetime

bmi_log = []


def convert_to_metric(weight, height, unit_system):
    """Convert weight and height to metric units."""
    if unit_system == 'metric':
        return weight, height / 100
    elif unit_system == 'imperial':
        weight_kg = weight * 0.453592
        height_m = height * 0.0254
        return weight_kg, height_m
    else:
        raise ValueError("Unit system must be 'metric' or 'imperial'")


def calculate_bmi(weight_kg, height_m):
    """Calculate BMI and category."""
    bmi = weight_kg / (height_m ** 2)
    
    if bmi < 18.5:
        category = 'Underweight'
    elif bmi < 25:
        category = 'Normal'
    elif bmi < 30:
        category = 'Overweight'
    else:
        category = 'Obese'
    
    return bmi, category


def add_bmi_entry(weight, height, unit_system='metric'):
    """Add a BMI entry with validation."""
    try:
        weight = float(weight)
        height = float(height)
        
        if unit_system == 'metric':
            if weight <= 0 or weight > 500:
                return "Error: Weight must be between 0 and 500 kg"
            if height <= 0 or height > 300:
                return "Error: Height must be between 0 and 300 cm"
        else:
            if weight <= 0 or weight > 1100:
                return "Error: Weight must be between 0 and 1100 lbs"
            if height <= 0 or height > 120:
                return "Error: Height must be between 0 and 120 inches"
        
        weight_kg, height_m = convert_to_metric(weight, height, unit_system)
        bmi, category = calculate_bmi(weight_kg, height_m)
        
        entry = {
            'timestamp': datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
            'weight_kg': weight_kg,
            'height_m': height_m,
            'bmi': bmi,
            'category': category
        }
        
        bmi_log.append(entry)
        return f"✓ Entry added: BMI = {bmi:.2f} ({category})"
    except ValueError as e:
        return f"Error: Invalid input - {e}"


def display_last_entries(n=10):
    """Display last n entries with ASCII trend bars."""
    if not bmi_log:
        return "No entries in log"
    
    entries = bmi_log[-n:]
    
    result = "\n" + "="*70 + "\n"
    result += f"Last {len(entries)} BMI Entries\n"
    result += "="*70 + "\n\n"
    result += f"{'Date':<20} {'BMI':<8} {'Category':<15} {'Trend'}\n"
    result += "-"*70 + "\n"
    
    for entry in entries:
        bmi = entry['bmi']
        bar_length = int((bmi - 15) * 2)
        bar_length = max(0, min(bar_length, 50))
        bar = "█" * bar_length
        result += f"{entry['timestamp']:<20} {bmi:>6.2f}  {entry['category']:<15} {bar}\n"
    
    result += "="*70 + "\n"
    return result


def export_to_csv():
    """Export log to CSV format."""
    if not bmi_log:
        return "No entries to export"
    
    csv = "timestamp,weight_kg,height_m,bmi,category\n"
    for entry in bmi_log:
        csv += f"{entry['timestamp']},{entry['weight_kg']:.2f},{entry['height_m']:.2f},"
        csv += f"{entry['bmi']:.2f},{entry['category']}\n"
    return csv

# Demo
print("=== BMI Logger Demo ===\n")
print(add_bmi_entry(70, 175, 'metric'))
print(add_bmi_entry(85, 180, 'metric'))
print(add_bmi_entry(154, 69, 'imperial'))
print(display_last_entries())
print("\nCSV Export (first 200 chars):")
print(export_to_csv()[:200] + "...")
```
