```python
# Student scores (nested data)
students = [
    {"name": "Amina", "assignment": 80, "test": 75, "exam": 90},
    {"name": "Sadiq", "assignment": 60, "test": 65, "exam": 70},
    {"name": "Zainab", "assignment": 90, "test": 88, "exam": 94},
    {"name": "Musa", "assignment": 50, "test": 55, "exam": 58}
]

totals = []
grades = []

# Function to assign letter grade
def get_grade(score):
    if score >= 90:
        return "A+"
    elif score >= 85:
        return "A"
    elif score >= 80:
        return "A-"
    elif score >= 75:
        return "B+"
    elif score >= 70:
        return "B"
    elif score >= 65:
        return "B-"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"


# Calculate totals and grades
for student in students:
    total = (
        student["assignment"] * 0.2 +
        student["test"] * 0.3 +
        student["exam"] * 0.5
    )

    grade = get_grade(total)

    student["total"] = round(total, 2)
    student["grade"] = grade

    totals.append(total)
    grades.append(grade)


# ---- Class Statistics ----

# Mean
mean_score = sum(totals) / len(totals)

# Median
sorted_totals = sorted(totals)
n = len(sorted_totals)

if n % 2 == 0:
    median_score = (sorted_totals[n//2 - 1] + sorted_totals[n//2]) / 2
else:
    median_score = sorted_totals[n//2]

# Grade distribution
grade_distribution = {}

for grade in grades:
    if grade in grade_distribution:
        grade_distribution[grade] += 1
    else:
        grade_distribution[grade] = 1


# ---- Text Report ----
print("\n===== CLASS GRADING REPORT =====\n")

for student in students:
    print(
        f"{student['name']}: "
        f"Total = {student['total']} | "
        f"Grade = {student['grade']}"
    )

print("\n--- Class Summary ---")
print(f"Mean Score: {round(mean_score, 2)}")
print(f"Median Score: {round(median_score, 2)}")

print("\nGrade Distribution:")
for grade, count in grade_distribution.items():
    print(f"{grade}: {count} student(s)")

print("\n===== END OF REPORT =====")
```
