```python
## Exercise 1: Robust Calculator



def safe_convert(value):
    """Safely convert input to numeric type (int or float)."""
    try:
        if isinstance(value, str) and '.' not in value:
            return int(value)
        return float(value)
    except (ValueError, TypeError):
        raise ValueError(f"Cannot convert '{value}' to a number")


def calculate(operand1, operator, operand2):
    """Perform calculation with error handling."""
    try:
        num1 = safe_convert(operand1)
        num2 = safe_convert(operand2)
        
        if operator not in ['+', '-', '*', '/']:
            raise ValueError(f"Invalid operator: {operator}")
        
        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            if num2 == 0:
                raise ZeroDivisionError("Cannot divide by zero")
            result = num1 / num2
        
        return result
    except ZeroDivisionError as e:
        return f"Error: {e}"
    except ValueError as e:
        return f"Error: {e}"
    except Exception as e:
        return f"Unexpected error: {e}"


def parse_expression(expression):
    """Parse and evaluate expression like '12.5 * 3'."""
    try:
        tokens = expression.split()
        if len(tokens) != 3:
            return "Error: Expression must be in format 'number operator number'"
        
        operand1, operator, operand2 = tokens
        return calculate(operand1, operator, operand2)
    except Exception as e:
        return f"Error parsing expression: {e}"

```

```python
print("=== Robust Calculator Demo ===\n")

# Take user inputs
value1 = input("Enter first value: ")
operation = input("Enter operation (+, -, *, /): ")
value2 = input("Enter second value: ")

# Perform calculation
result = calculate(value1, operation, value2)

print("\nResult:")
print(f"{value1} {operation} {value2} = {result}")
```

```python
print("\n=== Expression Mode ===")
expression = input("Enter an expression (e.g., 12.5 * 3): ")
print(f"Result: {parse_expression(expression)}")
```
