```python
#Data model
contacts = [
    {"name": "Amina Bello", "phone": "08031234567", "email": "amina@gmail.com"},
    {"name": "amina bello", "phone": "08031234567", "email": "amina.bello@gmail.com"},  # dup phone
    {"name": "John Musa", "phone": "07039876543", "email": "johnmusa@yahoo.com"},
    {"name": "Fatima Sule", "phone": "08123456789", "email": "fatima@gmail.com"},
    {"name": "FATIMA SULE", "phone": "08123456700", "email": "fatima@gmail.com"},       # dup email
]
```

```python
#Case insensitive substring search
def search_contacts(contacts, query):
    q = query.lower()
    return [
        c for c in contacts
        if q in c["name"].lower()
        or q in c["phone"].lower()
        or q in c["email"].lower()
    ]
```

```python
# Deduplication
def deduplicate_contacts(contacts):
    seen_phones = set()
    seen_emails = set()
    cleaned = []
    removed_log = []

    for c in contacts:
        phone = c["phone"]
        email = c["email"].lower()

        if phone in seen_phones or email in seen_emails:
            removed_log.append(c)
        else:
            cleaned.append(c)
            seen_phones.add(phone)
            seen_emails.add(email)

    return cleaned, removed_log
cleaned_contacts, duplicates = deduplicate_contacts(contacts)
```

```python
# Export cleaned contacts to CSV
import csv

def export_to_csv(filename, contacts):
    with open(filename, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["name", "phone", "email"])
        writer.writeheader()
        writer.writerows(contacts)
export_to_csv("cleaned_contacts.csv", cleaned_contacts)
export_to_csv("cleaned_contacts.csv", cleaned_contacts)
```

```python
# Duplicate removal log
def print_duplicate_log(duplicates):
    print("Removed Duplicates Log:")
    for d in duplicates:
        print(f"- {d['name']} | {d['phone']} | {d['email']}")

print_duplicate_log(duplicates)
```
